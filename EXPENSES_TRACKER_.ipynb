#EXPENSES TRACKER



```python
import json

try:
  with open("expenses.json", "r") as f:
    expenses = json.load(f)
except FileNotFoundError:
  expenses = []


while True :
  print("\n1. view expenses")
  print("2. add expenses")
  print("3. delete expenses")
  print("4. show total amount")
  print("5. exit")

  choice = input("\n choose 1/2/3/4/5:").strip()

  if choice == "1":
    if len(expenses) == 0:
      print("\nno expenses aded yet")
    else:
      for e in expenses:
        print(e["item"],"-",e["amount"])
  elif choice == "2":

    item = input("\nitem:").strip()

    try:
      amount = float(input("\namount: "))
    except ValueError:
      print("\ninvalid amount")
      continue

    expenses.append({"item": item,"amount":amount})
    print("\nexpenses added!")
  elif choice == "3":
    if len (expenses) == 0:
      print("Nothing to delete")
    else:
      for i, e in enumerate(expenses, start=1):
        print(i, ".", e["item"],"-",e["amount"] )
      try:

        num = int(input("number to delete:"))
      except ValueError:
        print("\ninvalid number")
        continue
      if num<1 or num>len(expenses):
        print("invalid number")
      else:
        expenses.pop(num -1)
        print ("\nexpenses removed!")

  elif choice =="4":
    total = 0
    for e in expenses:
      total += e["amount"]
    print(total)
  elif choice == "5":
    with open("expenses.json", "w") as f:
      json.dump(expenses, f)
    print("\n thank you!")
    break
  else :
    print("\ninvalid choice")
```
