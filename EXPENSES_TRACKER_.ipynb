#EXPENSES TRACKER



```python
expenses = []

while True :
  print("\n1. view expenses")
  print("2. add expenses")
  print("3. delete expenses")
  print("4. show total amount")
  print("5. exit")

  choice = input("\n choose 1/2/3/4/5:").strip()

  if choice == "1":
    if len(expenses) == 0:
      print("\nno expenses aded yet")
    else:
      for e in expenses:
        print(e["item"],"-",e["amount"])
  elif choice == "2":
    item = input("\nitem:")
    amount = float(input("amount: "))
    expenses.append({"item": item,"amount":amount})
    print("\nexpenses added!")
  elif choice == "3":
    print("\ndelete expense : No data yet!")
  elif choice =="4":
    total = 0
    for e in expenses:
      total += e["amount"]
    print(total)
  elif choice == "5":
    print("\n thank you!")
    break
  else :
    print("\ninvalid choice")
```
