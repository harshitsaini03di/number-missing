```python
import random

questions = [
    {
        "question": "fundamental rights is mention in which part of constitution ?",
        "options"  : ["part 1", "part 2", "part 3", "part 4"],
        "answer"  : "part 3"
    },
    {
        "question": "fuundament duties is mention is which part of constitution ?",
        "options" : [ "part 1 ", "part 4A", "part 3", " part 4"],
        "answer" : "part 4A"
    },
    {
         "question": "Which Article of the Constitution abolishes untouchability?",
        "options": ["Article 15", "Article 16", "Article 17", "Article 18"],
        "answer": "Article 17"
    },
    {
        "question": "Which Article did Dr. B.R. Ambedkar call the 'heart and soul' of the Constitution?",
        "options": ["Article 14", "Article 21", "Article 32", "Article 226"],
        "answer": "Article 32"
    },
    {
        "question": "Directive Principles of State Policy are given in which Part of the Constitution?",
        "options": ["Part III", "Part IV", "Part IV-A", "Part V"],
        "answer": "Part IV"
    },
    {
          "question": "Who was the Chairman of the Drafting Committee of the Constituent Assembly?",
        "options": ["Dr. Rajendra Prasad", "Dr. B.R. Ambedkar", "Jawaharlal Nehru", "Sardar Patel"],
        "answer": "Dr. B.R. Ambedkar"
    },
    {
        "question": "On which date did the Constitution of India come into force?",
        "options": ["15 August 1947", "26 November 1949", "26 January 1950", "26 January 1949"],
        "answer": "26 January 1950"
    },
    {
        "question": "Article 21 of the Constitution deals with which of the following?",
        "options": ["Right to equality", "Freedom of speech", "Protection of life and personal liberty", "Right to constitutional remedies"],
        "answer": "Protection of life and personal liberty"
    },
    {
         "question": "Who appoints the Governor of a State?",
        "options": ["Chief Minister", "President", "Prime Minister", "Chief Justice of India"],
        "answer": "President"
    },
    {
          "question": "What is the minimum age to be elected to the Lok Sabha?",
        "options": ["21 years", "25 years", "30 years", "35 years"],
        "answer": "25 years"
    },
    {
          "question": "What is the minimum age to be elected to the Lok Sabha?",
        "options": ["21 years", "25 years", "30 years", "35 years"],
        "answer": "25 years"
    },
    {
        "question": "Who is the constitutional head of the State of India?",
        "options": ["Prime Minister", "President", "Chief Justice", "Speaker"],
        "answer": "President"
    },
    {
       "question": "Which Article guarantees equality before the law?",
        "options": ["Article 14", "Article 19", "Article 21", "Article 32"],
        "answer": "Article 14"
    },
    {
        "question": "Which House of Parliament is called the Upper House?",
        "options": ["Lok Sabha", "Rajya Sabha", "Vidhan Sabha", "Gram Sabha"],
        "answer": "Rajya Sabha"
    }
]
random.shuffle(questions)
score = 0
labels = ["A", "B", "C", "D"]

for q in questions:
    print(q["question"])

    options = q["options"][:]
    random.shuffle(options)

    for label,option in zip(labels, options):
        print(f"{label}.{option}")

    answer = input("your answer is (A/B/C/D): ").strip().upper()

    while answer not in labels[:len(options)]:
      print ("INVALID answer. your option shpould be A, B, c, D " )
      answer = input("your answer is A/B/C/D: ").strip().upper()

    chosen = options[labels.index(answer)]

    if chosen == q["answer"]:
        print("Correct!\n")
        score += 1
    else:
      CORRECT_LABELS = labels[options.index(q["answer"])]
      print(f"Wrong. The correct answer was ({CORRECT_LABELS}. {q['answer']}).\n")

print(f"Your final score is: {score}/{len(questions)}")
print(f"percentage: {score/len(questions) * 100:.1f}%" )
```
