```python
questions = [
    {
        "question": "fundamental rights is mention in which part of constitution ?",
        "options"  : ["A. part 1", "B. part 2", "C. part 3", "D. part 4"],
        "answer"  : "C"
    },
    {
        "question": "fuundament duties is mention is which part of constitution ?",
        "options" : [ "A. part1", "B. part4A", "C. part3", "D. part4"],
        "answer" : "B"
    }
]

score = 0

for q in questions:
    print(q["question"])

    for option in q["options"]:
        print(option)

    answer = input("your answer is (A/B/C/D): ").strip().upper()

    if answer == q["answer"]:
        print("Correct!\n")
        score += 1
    else:
        print(f"Wrong. The correct answer was {q['answer']}.\n")

print(f"Your final score is: {score}/{len(questions)}")
```
