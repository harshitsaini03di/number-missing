```python
import random

words = ["python", "numpy", "pandas", "dictionary", "loop"]
scert = random.choice(words)
guessed = []
lives = 6

while lives > 0:

  display = ""
  for letter in scert:
    if letter in guessed:
      display += letter + " "
    else:
      display += "_ "
  if "_" not in display:
    print("\n" + display)
    print("\nYou won!")
    break

  print("\n" + display)
  print("lives left:",lives )

  guess = input("your guess:", ).strip().lower()
  guessed.append(guess)

  if guess not in scert :
    lives -= 1
    print("lives left:", lives)


  if lives == 0:

    print("\nyou lost")

```
