# number guessing game


```python
import random
Play = "yes"
while Play == "yes":

  secret =  random . randint (1, 100 )
  guess = int(input(" your guess :  "))
  attempt = 1

  while guess != secret and attempt < 7 :

    if guess < secret:
      print("too low ")
    elif guess > secret :
      print( "too high ")
    guess = int(input(" your guess :  "))
    attempt += 1
  if guess == secret :

    print ( "correct !")
    print ("attempt : ", attempt)
  else :
    print("game over ! , the secret:", secret )
  Play = input(" PLay again? (yes/no): ").strip().lower()
  print (repr(Play))
print ("thanks for playing !")
```
